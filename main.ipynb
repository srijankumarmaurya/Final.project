## 1. Load the dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

file_path = "happiness.csv"     

try:
    data = pd.read_csv(file_path)
    data.columns = data.columns.str.strip()   # remove extra spaces from column names
    print("Dataset loaded successfully!")
except Exception as e:
    print("Error loading file:", e)
```

## 2. Explore the dataset

```python
print("First 5 rows")
display(data.head())

print("\nNumber of rows and columns")
print("Rows:", data.shape[0], " Columns:", data.shape[1])

print("\nColumn names and data types")
print(data.dtypes)

print("\nSummary statistics")
display(data.describe())

print("\nMissing values in each column")
print(data.isnull().sum())
```

## 3. Clean the dataset

```python
print("Missing values in each column:")
print(data.isnull().sum())

num_cols = data.select_dtypes(include="number").columns
text_cols = data.select_dtypes(exclude="number").columns

# 1. Drop rows with missing values (shown on a copy)
print("\n1. Drop rows with missing values -> rows left:", len(data.dropna()), "of", len(data))

# 2. Fill with mean (shown on a copy)
data_mean = data.copy()
for col in num_cols:
    data_mean[col] = data_mean[col].fillna(data_mean[col].mean())
print("2. Fill numeric columns with mean -> missing numeric values left:", data_mean[num_cols].isnull().sum().sum())

# 3. Fill with median (shown on a copy)
data_median = data.copy()
for col in num_cols:
    data_median[col] = data_median[col].fillna(data_median[col].median())
print("3. Fill numeric columns with median -> missing numeric values left:", data_median[num_cols].isnull().sum().sum())

# 4. Fill with mode (shown on a copy)
data_mode = data.copy()
for col in data_mode.columns:
    mode = data_mode[col].mode()
    if not mode.empty:
        data_mode[col] = data_mode[col].fillna(mode.iloc[0])
print("4. Fill with mode -> missing values left:", data_mode.isnull().sum().sum())

# 5. Remove duplicates
before = len(data)
data = data.drop_duplicates().reset_index(drop=True)
print("\n5. Duplicate rows removed:", before - len(data))

# Final cleaning used in the rest of the notebook:
# numeric columns are filled with the mean, text columns with the mode
for col in num_cols:
    data[col] = data[col].fillna(data[col].mean())
for col in text_cols:
    mode = data[col].mode()
    if not mode.empty:
        data[col] = data[col].fillna(mode.iloc[0])

print("\nMissing values after cleaning:")
print(data.isnull().sum())
```

## 4. Analyze the dataset

```python
print("1. Happiest countries")
happiest = data.sort_values(by="Happiness Score", ascending=False).head()
display(happiest[["Country", "Happiness Score"]])

print("2. Least happy countries")
least_happy = data.sort_values(by="Happiness Score", ascending=True).head()
display(least_happy[["Country", "Happiness Score"]])

col = "Economy (GDP per Capita)"
print("3. Highest GDP per capita")
display(data.sort_values(by=col, ascending=False).head()[["Country", "Happiness Score", col]])
print("Lowest GDP per capita")
display(data.sort_values(by=col, ascending=True).head()[["Country", "Happiness Score", col]])

col = "Family"
print("4. Highest social support")
display(data.sort_values(by=col, ascending=False).head()[["Country", "Happiness Score", col]])
print("Lowest social support")
display(data.sort_values(by=col, ascending=True).head()[["Country", "Happiness Score", col]])

col = "Health (Life Expectancy)"
print("5. Highest life expectancy")
display(data.sort_values(by=col, ascending=False).head()[["Country", "Happiness Score", col]])
print("Lowest life expectancy")
display(data.sort_values(by=col, ascending=True).head()[["Country", "Happiness Score", col]])
```

## 5. Correlation analysis

```python
corr = data["Happiness Score"].corr(data["Economy (GDP per Capita)"])
print("Correlation between happiness score and GDP per capita:", round(corr, 3))

corr = data["Happiness Score"].corr(data["Family"])
print("Correlation between happiness score and social support:", round(corr, 3))

corr = data["Happiness Score"].corr(data["Health (Life Expectancy)"])
print("Correlation between happiness score and life expectancy:", round(corr, 3))
```

## 6. Visualize the dataset

```python
# 1. Bar chart of happiest countries
top10 = data.sort_values(by="Happiness Score", ascending=False).head(10)
plt.figure(figsize=(10, 6))
sns.barplot(data=top10, x="Country", y="Happiness Score")
plt.title("Happiest countries")
plt.xlabel("Country")
plt.ylabel("Happiness Score")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 2. Bar chart of least happy countries
bottom10 = data.sort_values(by="Happiness Score", ascending=True).head(10)
plt.figure(figsize=(10, 6))
sns.barplot(data=bottom10, x="Country", y="Happiness Score")
plt.title("Least happy countries")
plt.xlabel("Country")
plt.ylabel("Happiness Score")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 3. Scatter plot of Happiness Score vs GDP per Capita
plt.scatter(data["Economy (GDP per Capita)"], data["Happiness Score"])
plt.title("Happiness Score vs GDP per Capita")
plt.xlabel("GDP per Capita")
plt.ylabel("Happiness Score")
plt.show()

# 4. Scatter plot of Happiness Score vs Social Support
plt.scatter(data["Family"], data["Happiness Score"])
plt.title("Happiness Score vs Social Support")
plt.xlabel("Social Support")
plt.ylabel("Happiness Score")
plt.show()

# 5. Scatter plot of Happiness Score vs Life Expectancy
plt.scatter(data["Health (Life Expectancy)"], data["Happiness Score"])
plt.title("Happiness Score vs Life Expectancy")
plt.xlabel("Life Expectancy")
plt.ylabel("Happiness Score")
plt.show()

# 6. Heatmap of correlation between variables
corr_matrix = data.corr(numeric_only=True)
plt.figure(figsize=(9, 7))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```
